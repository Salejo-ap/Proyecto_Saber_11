## 0. Definición del Problema Real**

### **0.2 Problema concreto**

**Contexto**

En Colombia, los resultados de Saber 11 reflejan diferencias importantes entre establecimientos educativos, en un contexto marcado por brechas socioeconómicas, territoriales y de acceso a oportunidades de aprendizaje. Esta situación cobra especial relevancia ante los resultados recientes de PISA 2025: frente a 2015, la proporción de estudiantes colombianos con bajo desempeño aumentó en 12 puntos porcentuales en lectura y en 5 puntos porcentuales en matemáticas. Sin embargo, comparar colegios únicamente por sus puntajes puede ser injusto, porque no todos atienden estudiantes bajo las mismas condiciones.

**¿Qué problema se está abordando?**

El problema que se busca abordar es identificar qué sedes educativas presentan un desempeño mejor o peor de lo esperado según su contexto y cuáles muestran señales de deterioro frente a cohortes anteriores. Esto permitiría realizar comparaciones más justas y apoyar la priorización de acciones de acompañamiento y mejoramiento educativo.

Además, una lectura basada solo en puntajes absolutos puede ocultar tanto establecimientos que están obteniendo resultados destacables en contextos adversos como sedes cuyo desempeño se encuentra por debajo de lo esperable dadas sus condiciones. Identificar estas diferencias puede aportar una visión más útil para orientar decisiones de política y gestión educativa.

### **0.3 Stakeholder objetivo**

El stakeholder principal corresponde a las **Secretarías de Educación de las Entidades Territoriales Certificadas (ETC)**, particularmente a sus equipos de calidad educativa, planeación y analítica.

La herramienta podría apoyar la priorización de establecimientos que requieren acompañamiento, la identificación temprana de sedes cuya trayectoria muestra señales de deterioro y el reconocimiento de establecimientos que obtienen resultados superiores a los esperados bajo contextos similares. Estos últimos pueden constituir casos de interés para estudiar prácticas potencialmente replicables.

Sin embargo, los resultados de análisis se plantean como apoyo para la toma de decisiones y el diagnóstico, y no como mecanismo automático de sanción, clasificación institucional o asignación individual de recursos.

### **0.4 Costo de la inacción**

**¿Cuál es el impacto de no resolver este problema?**

Mantener la situación actual significa seguir interpretando el desempeño de los colegios principalmente a partir de sus puntajes finales. Esto puede hacer que una Secretaría de Educación detecte tarde una caída en el desempeño de una sede o concentre su atención únicamente en los establecimientos que históricamente han obtenido los resultados más bajos, sin distinguir si realmente están empeorando o si su resultado responde en buena medida al contexto en el que operan.

**¿Qué se pierde?**

Se pierde la posibilidad de hacer comparaciones más justas y útiles entre establecimientos. Un colegio con un puntaje bajo puede estar obteniendo resultados mejores de lo esperado dadas sus condiciones, mientras que otro con un puntaje aparentemente alto puede estar rindiendo por debajo de lo que cabría esperar. Sin esta perspectiva contextual, pueden pasar desapercibidas tanto las sedes que necesitan atención como aquellas que podrían convertirse en referentes de buenas prácticas.

### **0.5 Preguntas de negocio**

Para formular las preguntas de análisis se tomará como unidad de observación la **sede-jornada-cohorte**. Esta corresponde a una sede educativa específica, en una jornada determinada, para un grupo de estudiantes que presenta Saber 11 en el mismo periodo. De esta forma, el desempeño puede analizarse a nivel institucional y compararse a través del tiempo sin realizar predicciones sobre estudiantes individuales.

**1.** ¿Cuál es el desempeño esperado en Saber 11 de una sede-jornada-cohorte dadas sus características socioeconómicas, territoriales e institucionales?

**2.** ¿Qué sedes presentan resultados significativamente superiores o inferiores a los esperados frente a establecimientos que operan bajo contextos comparables?

**3.** ¿La trayectoria histórica del desempeño relativo de una sede permite anticipar un deterioro significativo en la siguiente cohorte?

**4.** ¿Existen grupos de establecimientos con contextos y trayectorias similares que permitan realizar comparaciones más pertinentes e identificar casos atípicos de interés?

### **0.6 Hipótesis iniciales**

A partir del problema planteado y de las preguntas de negocio, se proponen las siguientes hipótesis iniciales para orientar el análisis:

**H1. El contexto influye de manera importante en el desempeño de las sedes.**  
Se espera que las condiciones socioeconómicas, territoriales e institucionales expliquen una parte relevante de los resultados en Saber 11. Sin embargo, también se espera encontrar sedes que obtengan resultados claramente mejores o peores de lo que cabría esperar bajo condiciones similares.

**H2. La trayectoria previa de una sede permite anticipar cambios en su desempeño.**  
Se espera que el comportamiento observado en cohortes anteriores, especialmente su tendencia y estabilidad, aporte información útil para identificar sedes con riesgo de deterioro en periodos posteriores.

**H3. Los modelos de Machine Learning pueden capturar patrones que no son evidentes mediante reglas simples.**  
Se espera que algunos modelos no lineales logren representar mejor las interacciones entre contexto y desempeño que un promedio histórico o un modelo lineal básico. Esta hipótesis deberá comprobarse experimentalmente y se considerará rechazada si los modelos más complejos no muestran una mejora relevante.

### **0.7 Formulación analítica y justificación del uso de Machine Learning**

A partir de las preguntas de negocio, el problema se abordará mediante dos tareas principales de Machine Learning, utilizando como unidad de análisis la sede-jornada-cohorte definida previamente.

**Problema 1 - Regresión: estimación del desempeño esperado.**  
Se buscará estimar el puntaje global promedio esperado de una sede dadas sus condiciones socioeconómicas, territoriales e institucionales. La diferencia entre el puntaje observado y el estimado permitirá identificar sedes cuyo desempeño se encuentre por encima o por debajo de lo esperado para su contexto. Para esta tarea podrán compararse modelos como Regresión Lineal, SVR, Árboles de Decisión, Random Forest y Gradient Boosting.

**Problema 2 - Clasificación: alerta de deterioro futuro.**  
Se buscará determinar si una sede presenta riesgo de experimentar un deterioro significativo de su desempeño relativo en la siguiente cohorte, utilizando únicamente información disponible en periodos anteriores. Para ello podrán emplearse variables relacionadas con su trayectoria reciente, tendencia y estabilidad, y compararse modelos como Regresión Logística, SVM, Árboles de Decisión, Random Forest y Gradient Boosting.

**¿Por qué usar Machine Learning?**

El uso de Machine Learning se justifica porque el objetivo no consiste únicamente en describir puntajes históricos o construir un ranking de establecimientos. Un dashboard podría mostrar qué sedes obtuvieron mejores o peores resultados, pero no estimaría qué desempeño sería razonable esperar dadas sus condiciones ni permitiría anticipar deterioros futuros a partir de múltiples patrones simultáneos.

Sin embargo, no se asumirá que los modelos más complejos son necesariamente superiores. Su aporte deberá demostrarse comparándolos con alternativas simples, como el promedio histórico de la sede o modelos lineales. Si no se obtiene una mejora relevante bajo validación temporal, se privilegiará la alternativa más simple e interpretable.

```python
# Variables que se conservarán en los microdatos históricos

variables_microdatos = [
    # Identificación
    "periodo",
    "estu_consecutivo",
    "cole_cod_dane_establecimiento",
    "cole_cod_dane_sede",
    "cole_codigo_icfes",
    "cole_jornada",

    # Contexto institucional y territorial
    "cole_area_ubicacion",
    "cole_calendario",
    "cole_caracter",
    "cole_naturaleza",
    "cole_depto_ubicacion",
    "cole_mcpio_ubicacion",

    # Contexto del estudiante y hogar
    "estu_genero",
    "fami_educacionmadre",
    "fami_educacionpadre",
    "fami_estratovivienda",
    "fami_tieneinternet",
    "fami_tienecomputador",
    "fami_tienelavadora",
    "fami_tieneautomovil",
    "fami_personashogar",

    # Resultados
    "punt_global",
    "punt_matematicas",
    "punt_lectura_critica",
    "punt_c_naturales",
    "punt_sociales_ciudadanas",
    "punt_ingles"
]

# Verificar que ninguna variable necesaria haya desaparecido
faltantes = [
    v for v in variables_microdatos
    if v not in disponibles
]

if faltantes:
    raise ValueError(
        f"Variables no disponibles en la fuente: {faltantes}"
    )

# Variables numéricas de resultados
columnas_puntaje = [
    "punt_global",
    "punt_matematicas",
    "punt_lectura_critica",
    "punt_c_naturales",
    "punt_sociales_ciudadanas",
    "punt_ingles"
]

# Identificadores que deben tratarse como texto
columnas_id = [
    "periodo",
    "estu_consecutivo",
    "cole_cod_dane_establecimiento",
    "cole_cod_dane_sede",
    "cole_codigo_icfes"
]

# Unidad de análisis que se construirá posteriormente
llave_unidad = [
    "periodo",
    "cole_cod_dane_sede",
    "cole_jornada"
]

print(f"Variables seleccionadas: {len(variables_microdatos)}")
print("Unidad final:", " + ".join(llave_unidad))
```

```python
# Directorios de trabajo reproducibles

from pathlib import Path

RUTA_BASE = Path("/content/Proyecto_ML_Saber11")

RUTA_MICRODATOS = (
    RUTA_BASE / "data" / "microdatos_limpios"
)

RUTA_DATAICFES = (
    RUTA_BASE / "data" / "dataicfes_original"
)

RUTA_MICRODATOS.mkdir(parents=True, exist_ok=True)
RUTA_DATAICFES.mkdir(parents=True, exist_ok=True)

print("Directorio de trabajo:", RUTA_BASE)
```

```python
# Endpoint de exportación masiva de Socrata
EXPORT_URL = (
    f"https://www.datos.gov.co/api/v3/views/"
    f"{DATASET_ID}/export.csv"
)


def descargar_cohorte(periodo_fuente):
    """
    Descarga una cohorte desde Datos Abiertos, elimina duplicados
    exactos en el servidor y realiza validaciones básicas.
    """

    columnas = ", ".join(variables_microdatos)

    query = f"""
        SELECT DISTINCT {columnas}
        WHERE periodo = '{periodo_fuente}'
        AND estu_estudiante = 'ESTUDIANTE'
    """

    response = requests.post(
        EXPORT_URL,
        json={
            "query": query,
            "orderingSpecifier": "discard"
        },
        timeout=600
    )

    response.raise_for_status()

    df = pd.read_csv(
    StringIO(response.text),
    dtype="string"
    )

    # Normalizar nombres de columnas
    df.columns = (
        df.columns
        .str.strip()
        .str.lower()
    )

    # Comprobar que llegaron todas las variables esperadas
    faltantes = [
        col for col in variables_microdatos
        if col not in df.columns
    ]

    if faltantes:
        raise ValueError(
            f"{periodo_fuente}: faltan columnas {faltantes}"
        )

    # Conservar código original de la fuente
    df["periodo_fuente"] = (
        df["periodo"]
        .astype(str)
    )

    # Normalizar la secuencia temporal
    df["periodo"] = (
        df["periodo_fuente"]
        .replace(mapa_periodos)
    )

    # Identificadores como texto
    for col in columnas_id:
        if col in df.columns:
            df[col] = (
                df[col]
                .astype("string")
                .str.replace(r"\.0$", "", regex=True)
            )

    # Puntajes como variables numéricas
    for col in columnas_puntaje:
        df[col] = pd.to_numeric(
            df[col],
            errors="coerce"
        )

    # Validar que quede una observación por estudiante
    n_registros = len(df)
    n_estudiantes = df["estu_consecutivo"].nunique()

    if n_registros != n_estudiantes:
        raise ValueError(
            f"{periodo_fuente}: quedaron "
            f"{n_registros:,} registros para "
            f"{n_estudiantes:,} estudiantes únicos."
        )

    print(
        f"{periodo_fuente} → "
        f"{df['periodo'].iloc[0]} | "
        f"{n_registros:,} estudiantes"
    )

    return df
```

```python
# Construcción del histórico desde Datos Abiertos

%pip install pyarrow

import gc

for i, periodo_fuente in enumerate(periodos_api, start=1):

    print(
        f"\n[{i}/{len(periodos_api)}] "
        f"Procesando periodo {periodo_fuente}"
    )

    # Archivo definitivo de la cohorte
    archivo_final = (
        RUTA_MICRODATOS
        / f"saber11_{periodo_fuente}.parquet"
    )

    # Archivo temporal mientras se completa el guardado
    archivo_temporal = (
        RUTA_MICRODATOS
        / f"saber11_{periodo_fuente}_temp.parquet"
    )

    # Si ya fue procesado correctamente, no se vuelve a descargar
    if archivo_final.exists():
        print("Ya existe → se omite.")
        continue

    # Eliminar temporal incompleto de una ejecución anterior
    if archivo_temporal.exists():
        archivo_temporal.unlink()

    try:
        # Descargar y validar cohorte
        df_cohorte = descargar_cohorte(periodo_fuente)

        # Guardar primero como archivo temporal
        df_cohorte.to_parquet(
            archivo_temporal,
            index=False
        )

        # Solo cuando termina correctamente se convierte en definitivo
        archivo_temporal.replace(archivo_final)

        print(
            f"Guardado correctamente → "
            f"{archivo_final.name}"
        )

        # Liberar memoria antes de la siguiente cohorte
        del df_cohorte
        gc.collect()

    except Exception as e:
        print(
            f"ERROR en el periodo {periodo_fuente}: {e}"
        )

        # No continuar silenciosamente si una cohorte falla
        raise
```

```python
# Procesar una cohorte histórica de DataIcfes

def procesar_cohorte_dataicfes(periodo, archivo):

    columnas_lectura = (
        variables_microdatos
        + ["estu_estudiante"]
    )

    df = pd.read_csv(
        archivo,
        sep=";",
        encoding="utf-8-sig",
        dtype="string",
        usecols=columnas_lectura,
        low_memory=False
    )

    # Normalizar nombres
    df.columns = (
        df.columns
        .str.strip()
        .str.lower()
    )

    # Conservar estudiantes vinculados a instituciones
    df["estu_estudiante"] = (
        df["estu_estudiante"]
        .str.strip()
        .str.upper()
    )

    df = df[
        df["estu_estudiante"] == "ESTUDIANTE"
    ].drop(
        columns="estu_estudiante"
    ).copy()

    # Eliminar posibles duplicados exactos
    df.drop_duplicates(
        inplace=True
    )

    # Estandarizar identificadores
    for col in columnas_id:
        df[col] = (
            df[col]
            .astype("string")
            .str.strip()
            .str.replace(r"\.0$", "", regex=True)
        )

    # Validar periodo
    if set(df["periodo"].dropna().unique()) != {periodo}:
        raise ValueError(
            f"{periodo}: el archivo contiene "
            "periodos inesperados."
        )

    # Conservar código de la fuente
    df["periodo_fuente"] = df["periodo"]

    # Puntajes numéricos
    for col in columnas_puntaje:
        df[col] = pd.to_numeric(
            df[col],
            errors="coerce"
        )

    # Validar unicidad del estudiante
    if len(df) != df["estu_consecutivo"].nunique():
        raise ValueError(
            f"{periodo}: existen estudiantes "
            "con más de un registro diferente."
        )

    # Igualar estructura con Datos Abiertos
    return df[
        variables_microdatos
        + ["periodo_fuente"]
    ]
```

### **1.1 EDA con Narrativa**

Se explora el comportamiento de las sedes-jornada-cohorte para comprender la distribución del desempeño, su evolución temporal y su relación con características institucionales y socioeconómicas.

Cada análisis se plantea a partir de una pregunta concreta del problema, de modo que las estadísticas y visualizaciones permitan identificar patrones, diferencias de contexto, sesgos y limitaciones relevantes antes de avanzar al modelado.

```python
# Construcción de la base de análisis a nivel sede-jornada-cohorte

bases_cohorte = []

columnas_eda_base = [
    "periodo",
    "estu_consecutivo",
    "cole_cod_dane_sede",
    "cole_jornada",
    "cole_area_ubicacion",
    "cole_calendario",
    "cole_caracter",
    "cole_naturaleza",
    "cole_depto_ubicacion",
    "cole_mcpio_ubicacion",
    "punt_global",
    "punt_matematicas",
    "punt_lectura_critica",
    "punt_c_naturales",
    "punt_sociales_ciudadanas",
    "punt_ingles"
]

for _, fila in catalogo_historico.iterrows():

    df = pd.read_parquet(
        fila["archivo"],
        columns=columnas_eda_base
    )

    # Una observación por sede-jornada-cohorte
    cohorte = (
        df.groupby(
            [
                "periodo",
                "cole_cod_dane_sede",
                "cole_jornada"
            ],
            dropna=False
        )
        .agg(
            n_estudiantes=(
                "estu_consecutivo",
                "nunique"
            ),

            punt_global_promedio=(
                "punt_global",
                "mean"
            ),

            punt_global_mediana=(
                "punt_global",
                "median"
            ),

            punt_global_std=(
                "punt_global",
                "std"
            ),

            matematicas_promedio=(
                "punt_matematicas",
                "mean"
            ),

            lectura_promedio=(
                "punt_lectura_critica",
                "mean"
            ),

            ciencias_promedio=(
                "punt_c_naturales",
                "mean"
            ),

            sociales_promedio=(
                "punt_sociales_ciudadanas",
                "mean"
            ),

            ingles_promedio=(
                "punt_ingles",
                "mean"
            ),

            area=(
                "cole_area_ubicacion",
                "first"
            ),

            calendario=(
                "cole_calendario",
                "first"
            ),

            caracter=(
                "cole_caracter",
                "first"
            ),

            naturaleza=(
                "cole_naturaleza",
                "first"
            ),

            departamento=(
                "cole_depto_ubicacion",
                "first"
            ),

            municipio=(
                "cole_mcpio_ubicacion",
                "first"
            )
        )
        .reset_index()
    )

    bases_cohorte.append(cohorte)

base_eda = pd.concat(
    bases_cohorte,
    ignore_index=True
)

# Orden temporal
base_eda["periodo"] = (
    base_eda["periodo"]
    .astype(str)
)

base_eda = base_eda.sort_values(
    "periodo"
).reset_index(drop=True)

print("Dimensiones de la base EDA:", base_eda.shape)
print(
    "Número de cohortes:",
    base_eda["periodo"].nunique()
)

display(base_eda.head())
```

```python
# Cobertura y tamaño de las unidades de análisis por periodo

resumen_cobertura = (
    base_eda
    .groupby("periodo")
    .agg(
        estudiantes=(
            "n_estudiantes",
            "sum"
        ),

        sedes=(
            "cole_cod_dane_sede",
            "nunique"
        ),

        unidades_sede_jornada=(
            "cole_cod_dane_sede",
            "size"
        ),

        mediana_estudiantes_unidad=(
            "n_estudiantes",
            "median"
        ),

        promedio_estudiantes_unidad=(
            "n_estudiantes",
            "mean"
        )
    )
    .reset_index()
)

# Unidades potencialmente inestables por bajo número de estudiantes
prop_pequenas = (
    base_eda
    .assign(
        menos_10=lambda x:
        x["n_estudiantes"] < 10
    )
    .groupby("periodo")["menos_10"]
    .mean()
    .mul(100)
    .reset_index(
        name="pct_unidades_menos_10"
    )
)

resumen_cobertura = resumen_cobertura.merge(
    prop_pequenas,
    on="periodo",
    how="left"
)

resumen_cobertura[
    "promedio_estudiantes_unidad"
] = resumen_cobertura[
    "promedio_estudiantes_unidad"
].round(1)

resumen_cobertura[
    "mediana_estudiantes_unidad"
] = resumen_cobertura[
    "mediana_estudiantes_unidad"
].round(1)

resumen_cobertura[
    "pct_unidades_menos_10"
] = resumen_cobertura[
    "pct_unidades_menos_10"
].round(1)

display(resumen_cobertura)
```

```python
# Distribución del desempeño entre unidades sede-jornada-cohorte

def resumir_desempeno(df, nombre):

    serie = df["punt_global_promedio"].dropna()

    return {
        "muestra": nombre,
        "n_unidades": len(serie),
        "media": serie.mean(),
        "mediana": serie.median(),
        "desv_estandar": serie.std(),
        "q1": serie.quantile(0.25),
        "q3": serie.quantile(0.75),
        "iqr": serie.quantile(0.75) - serie.quantile(0.25),
        "minimo": serie.min(),
        "maximo": serie.max(),
        "asimetria": serie.skew()
    }


resumen_desempeno = pd.DataFrame([
    resumir_desempeno(
        base_eda,
        "Todas las unidades"
    ),

    resumir_desempeno(
        base_eda[
            base_eda["n_estudiantes"] >= 10
        ],
        "Unidades con ≥10 estudiantes"
    )
])

columnas_redondear = [
    "media",
    "mediana",
    "desv_estandar",
    "q1",
    "q3",
    "iqr",
    "minimo",
    "maximo",
    "asimetria"
]

resumen_desempeno[columnas_redondear] = (
    resumen_desempeno[columnas_redondear]
    .round(2)
)

display(resumen_desempeno)
```

```python
import numpy as np
import seaborn as sns

# Funciones auxiliares

def normalizar_texto(serie):
    return (
        serie.astype("string")
        .str.strip()
        .str.upper()
        .str.normalize("NFKD")
        .str.encode("ascii", errors="ignore")
        .str.decode("utf-8")
    )


def convertir_si_no(serie):
    texto = normalizar_texto(serie)

    return texto.map({
        "SI": 1.0,
        "NO": 0.0
    })


# Construcción de indicadores por sede-jornada-cohorte

bases_socio_eda = []

columnas_socio = [
    "periodo",
    "cole_cod_dane_sede",
    "cole_jornada",
    "estu_genero",
    "fami_educacionmadre",
    "fami_educacionpadre",
    "fami_estratovivienda",
    "fami_tieneinternet",
    "fami_tienecomputador",
    "fami_tienelavadora",
    "fami_tieneautomovil",
    "fami_personashogar"
]

for _, fila in catalogo_historico.iterrows():

    df = pd.read_parquet(
        fila["archivo"],
        columns=columnas_socio
    )

    # Acceso a bienes y conectividad
    df["internet"] = convertir_si_no(
        df["fami_tieneinternet"]
    )

    df["computador"] = convertir_si_no(
        df["fami_tienecomputador"]
    )

    df["lavadora"] = convertir_si_no(
        df["fami_tienelavadora"]
    )

    df["automovil"] = convertir_si_no(
        df["fami_tieneautomovil"]
    )

    # Estratos 1 y 2
    estrato = normalizar_texto(
        df["fami_estratovivienda"]
    )

    df["estrato_1_2"] = np.where(
        estrato.isin(["ESTRATO 1", "ESTRATO 2"]),
        1.0,
        np.where(
            estrato.isin([
                "ESTRATO 3",
                "ESTRATO 4",
                "ESTRATO 5",
                "ESTRATO 6"
            ]),
            0.0,
            np.nan
        )
    )

    # Género
    genero = normalizar_texto(
        df["estu_genero"]
    )

    df["mujer"] = genero.map({
        "F": 1.0,
        "M": 0.0
    })

    # Educación superior de madre y padre
    for origen, destino in [
        ("fami_educacionmadre", "madre_superior"),
        ("fami_educacionpadre", "padre_superior")
    ]:

        educacion = normalizar_texto(df[origen])

        df[destino] = np.where(
            educacion.str.contains(
                "TECNIC|TECNOLOG|PROFESIONAL|POSTGRADO",
                regex=True,
                na=False
            ),
            1.0,
            np.where(
                educacion.notna()
                & ~educacion.str.contains(
                    "NO SABE|NO APLICA",
                    regex=True,
                    na=False
                ),
                0.0,
                np.nan
            )
        )

    # Número de personas en el hogar
    df["personas_hogar"] = pd.to_numeric(
        df["fami_personashogar"],
        errors="coerce"
    )

    socio = (
        df.groupby(
            [
                "periodo",
                "cole_cod_dane_sede",
                "cole_jornada"
            ],
            dropna=False
        )
        .agg(
            prop_internet=("internet", "mean"),
            prop_computador=("computador", "mean"),
            prop_lavadora=("lavadora", "mean"),
            prop_automovil=("automovil", "mean"),
            prop_estrato_1_2=("estrato_1_2", "mean"),
            prop_mujeres=("mujer", "mean"),
            prop_madre_superior=("madre_superior", "mean"),
            prop_padre_superior=("padre_superior", "mean"),
            promedio_personas_hogar=("personas_hogar", "mean")
        )
        .reset_index()
    )

    bases_socio_eda.append(socio)


socio_eda_final = pd.concat(
    bases_socio_eda,
    ignore_index=True
)

print(
    "Unidades con indicadores socioeconómicos:",
    len(socio_eda_final)
)

display(socio_eda_final.head())
```

### **1.2 Preparación de Datos e Ingeniería de Características**

Una parte del preprocesamiento ya fue realizada durante la consolidación histórica de datos, donde se estandarizaron las fuentes, periodos, tipos de variables y registros individuales. Sin embargo, aquí el objetivo cambia: transformar ese histórico limpio en un conjunto de datos directamente utilizable para el modelado a nivel de sede-jornada-cohorte.

A partir de los hallazgos del EDA se definen criterios de estabilidad, tratamiento de valores faltantes e ingeniería de características institucionales y socioeconómicas. Las transformaciones dependientes del modelo, como la estandarización de variables numéricas o la codificación de categorías, se aplicarán posteriormente dentro de los *pipelines* de entrenamiento específicos.

**Construcción del conjunto de modelado**

Se parte del histórico ya depurado y de la unidad sede-jornada-cohorte construida durante el EDA. A esta base se incorporan los indicadores socioeconómicos agregados, conservando únicamente variables disponibles para describir el contexto institucional y familiar.

Los puntajes de las pruebas específicas se mantienen separados del conjunto de predictores, ya que el puntaje global se construye a partir de ellos y utilizarlos simultáneamente produciría fuga de información.

```python
# Integración de la base definitiva de modelado

llave = ["periodo", "cole_cod_dane_sede", "cole_jornada"]

variables_contexto = [
    "n_estudiantes",
    "area", "calendario", "caracter", "naturaleza",
    "departamento", "municipio",
    "prop_internet", "prop_computador", "prop_lavadora",
    "prop_automovil", "prop_estrato_1_2",
    "prop_madre_superior", "prop_padre_superior",
    "prop_mujeres"
]

base_modelo = (
    base_eda
    .merge(
        socio_eda_final,
        on=llave,
        how="left",
        validate="one_to_one"
    )
)

base_modelo["anio"] = (
    base_modelo["periodo"].astype(str).str[:4].astype(int)
)

base_modelo["aplicacion"] = (
    base_modelo["periodo"].astype(str).str[-1].astype(int)
)

base_modelo = base_modelo[
    llave
    + ["anio", "aplicacion"]
    + variables_contexto
    + ["punt_global_promedio"]
].copy()

# Auditoría previa al tratamiento
faltantes = (
    base_modelo.isna()
    .mean()
    .mul(100)
    .round(1)
    .sort_values(ascending=False)
)

print("Dimensiones:", base_modelo.shape)
print("Duplicados en la llave:", base_modelo.duplicated(llave).sum())
print("Unidades con <10 estudiantes:",
      round((base_modelo["n_estudiantes"] < 10).mean() * 100, 1), "%")

display(
    faltantes[faltantes > 0]
    .to_frame("% faltantes")
)
```

**Limpieza orientada al modelado**

La base consolidada no presenta duplicados y conserva únicamente unidades con al menos 10 estudiantes por sede-jornada-cohorte, reduciendo la inestabilidad de promedios calculados con grupos muy pequeños. No se eliminan valores extremos de desempeño, ya que identificar resultados atípicos constituye uno de los objetivos del proyecto.

Los valores faltantes en variables categóricas se representan mediante la categoría "Desconocido", preservando las observaciones sin asumir una categoría existente. Los faltantes numéricos se imputarán posteriormente mediante la mediana dentro de los *pipelines* de modelado, utilizando únicamente los datos de entrenamiento.

```python
# Tratamiento de faltantes categóricos

variables_categoricas = [
    "area",
    "calendario",
    "caracter",
    "naturaleza",
    "departamento",
    "municipio"
]

base_modelo[variables_categoricas] = (
    base_modelo[variables_categoricas]
    .fillna("Desconocido")
)
```

```python
# Criterios mínimos de calidad para modelado

n_inicial = len(base_modelo)

base_modelo = (
    base_modelo
    .dropna(subset=["cole_jornada", "punt_global_promedio"])
    .query("n_estudiantes >= 10")
    .reset_index(drop=True)
)

print("Unidades iniciales:", n_inicial)
print("Unidades finales:", len(base_modelo))
print("Retención:", f"{len(base_modelo) / n_inicial:.1%}")
print("Duplicados:", base_modelo.duplicated(llave).sum())
```

**Ingeniería de características**

A partir de la base ya depurada se crean nuevas variables que resumen dos dimensiones relevantes para el modelado: el **tamaño de la cohorte** y la **trayectoria histórica de cada sede-jornada**.

El número de estudiantes se transforma mediante `log(1 + n_estudiantes)` para reducir su asimetría y evitar que las cohortes muy grandes dominen la escala de esta variable. Además, se construyen tres características temporales: **puntaje previo**, que recupera el resultado de la cohorte comparable inmediatamente anterior; **variación previa**, que mide el cambio entre las dos últimas cohortes disponibles; y **cohortes previas**, que indica cuántos antecedentes históricos existen para esa sede-jornada.

La búsqueda de antecedentes se realiza dentro de la misma **sede, jornada y tipo de aplicación**. Esto es importante porque el EDA mostró que las aplicaciones 1 y 2 representan poblaciones institucionales diferentes; por ejemplo, para una observación de `2022-2` se busca primero el antecedente de otra Aplicación 2 y no simplemente la cohorte cronológicamente anterior. De esta forma, las variables históricas comparan contextos más equivalentes y utilizan únicamente información disponible antes de cada observación.

Los valores faltantes que aparecen en estas variables temporales son estructurales: una sede sin historia previa no puede tener `puntaje_previo`, y se requieren al menos dos antecedentes para calcular `variacion_previa`.

```python
# Características de tamaño y trayectoria histórica

base_modelo = base_modelo.sort_values(
    ["cole_cod_dane_sede", "cole_jornada", "aplicacion", "periodo"]
).copy()

grupo = base_modelo.groupby(
    ["cole_cod_dane_sede", "cole_jornada", "aplicacion"]
)

base_modelo["log_n_estudiantes"] = np.log1p(
    base_modelo["n_estudiantes"]
)

base_modelo["puntaje_previo"] = grupo[
    "punt_global_promedio"
].shift(1)

base_modelo["variacion_previa"] = (
    grupo["punt_global_promedio"].shift(1)
    - grupo["punt_global_promedio"].shift(2)
)

base_modelo["cohortes_previas"] = grupo.cumcount()

display(
    base_modelo[
        llave + [
            "n_estudiantes",
            "log_n_estudiantes",
            "puntaje_previo",
            "variacion_previa",
            "cohortes_previas"
        ]
    ].head(10)
)
```

**Preparación para el modelado**

Las variables se separan entre numéricas y categóricas para aplicar posteriormente el preprocesamiento adecuado dentro de cada *pipeline*. Las variables numéricas faltantes se imputarán mediante la mediana y, cuando el algoritmo lo requiera, se estandarizarán; las variables categóricas se codificarán mediante *One-Hot Encoding*. Estos parámetros se ajustarán exclusivamente con los datos de entrenamiento.

```python
# Variables disponibles para el modelado

numericas_contexto = [
    "log_n_estudiantes",
    "prop_internet", "prop_computador",
    "prop_lavadora", "prop_automovil",
    "prop_estrato_1_2",
    "prop_madre_superior", "prop_padre_superior",
    "prop_mujeres"
]

categoricas_contexto = [
    "area", "calendario", "caracter",
    "naturaleza", "departamento", "municipio",
    "aplicacion"
]

variables_temporales = [
    "puntaje_previo",
    "variacion_previa",
    "cohortes_previas"
]

print("Variables de contexto:", len(numericas_contexto) + len(categoricas_contexto))
print("Variables temporales:", len(variables_temporales))
```
